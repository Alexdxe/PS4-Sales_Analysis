## <span style=color:blue> Part 1: Imports </span>

```python
#General Packages:
import pandas as pd
import numpy as np

#Visualizations:
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#Dataset:
game_df = pd.read_csv('data/Updated_PS4_Game_Sales.csv', encoding = 'latin1')
```

```python
game_df.head()
```

## <span style=color:blue> Part 2: Visualizations</span>

### <span style=color:teal>Pie Chart for Total Sales:</span>

```python
#Finding sum of each region:
na_sum = game_df['North America'].sum()
eu_sum = game_df['Europe'].sum()
jp_sum = game_df['Japan'].sum()
rest_sum = game_df['Rest of World'].sum()

#Compiling them into a series:
data = [na_sum, eu_sum, jp_sum, rest_sum] #Making a list
game_s = pd.Series(data, index = ['NA', 'Europe', 'Japan', 'Rest']) #Compiling it into a series.
```

```python
#Making the plot: 
px.pie(
    game_s,
    values = game_s.values,
    names = game_s.index,
    title = 'Breakdown of Total PS4 Sales Across Regions',
    template = 'seaborn',
    color_discrete_sequence = px.colors.sequential.Blues[2:]
)
    
```

#### Notes: 
1) The PS4's biggest market (in terms of games) is Europe, accounting for 43% of it's total software sales.
2) Despite Sony being a Japanese company, it only accounted for 5% of Sony's overall game sales.

### <span style=color:teal>Distribution of Game Genres:</span>

```python
#Finding the value counts:
game_counts = game_df['Genre'].value_counts()
```

```python
#Plotting the Game Genres: 
px.bar(
    game_counts,
    x = game_counts.index,
    y = game_counts.values,
    title = 'Distribution of PS4 Game Genres',
    labels = 
        {'x' : 'Game Genres',
         'y' : 'Counts of Genre'},
    color = game_counts.values,
    color_continuous_scale="teal")
```

#### Notes: 
1) Action games tend to dominate the majority of the PS4's library.
2) Party, Visual novels, MMOs, and Puzzles are the least popular genres.

### <span style=color:teal>Distribution of Game Release Years:</span>

```python
#Finding the value counts:
year_counts = game_df['Year'].value_counts()
```

```python
#Plotting the Years:
px.bar(
    year_counts,
    x = year_counts.index,
    y = year_counts.values,
    title = 'Distribution of Games by Release Date',
    labels = 
    {'x' : 'Year',
     'y' : 'Game Count'}
)
```

#### Notes: 
1) The PS4 had a steady rise in games from 2014 -> 2017, with a sharp decline occurring in 2018.
2) The decline was likely due to developers working on other titles (after releasing games earlier in the year) or games being unaccounted for.
